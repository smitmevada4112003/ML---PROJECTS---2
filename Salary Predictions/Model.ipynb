```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression # Sai kit learn.
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r"D:\Machine learnings datasets\Salary_dataset.csv")
df
```

```python
df.head()
```

```python
df.columns   # X = 2D, y = 1D
```

```python
X = df[["YearsExperience"]]  # input 2D
Y = df["Salary"]
# 1
print(X.head())
# print(Y.head())  # input and Output, input - Single Values.
# Degree Age
```

```python
# Data Split - X train x test Y train y test - train Split.
# test_Size = 0.20 // 20, 80 - training, random_State = 42 // integer fun / references.

X_train,X_test,Y_train,Y_test = train_test_split(

    X,Y, test_size = 0.20, random_state = 42
)
```

```python
model = LinearRegression()  # Model Constructors Same linear Class initalizations.
```

```python
model.fit(X_train,Y_train)  # input and Output.
```

```python
v = model.predict(X_test)
Scores = r2_score(v,Y_test)
print(Scores)

# R^2 Scores = 0 to 1 = 1 Perfact 0.8 to 0.96
```

```python
# y = mx + c // 9449(2) + 24k = y
# Output = Slope (input) + Constant
print(model.coef_[0])  # experience 1 = 9k.
print(model.intercept_)  # rapido = 50Rs.
```

```python
Results = model.predict([[6]])
print(Results[0])
```

```python
plt.scatter(X,Y,color = "blue")
plt.plot(X,model.predict(X), color= "red")
plt.xlabel("exp")
plt.ylabel("Salary")
plt.title("expe and Salary")
plt.show()
```
